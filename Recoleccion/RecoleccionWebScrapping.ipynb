```python
#Libreria para aplicar funciones reges
import re
#Libreria para obtener el html de una url
from urllib.request import urlopen
#Libreria para webscrapping
from bs4 import BeautifulSoup
import pandas as pd

from google.colab import files

import json
```

```python
#############################################################Vamos a intentar, crear funciones y guardar en una lista de diccionarios los primeros 60 pisos de pisos.com, de esta manera,
# practicamos todo para enviarlo a csv:

def recoger_datos_web(url_web, lista_diccionarios):   #Le pasamos un url de pg web y nos devuelve todos los datos de los 30 pisos que encontramos en la pg
    html = urlopen(url_web, timeout=30)

    # Parseamos el html.
    soup = BeautifulSoup(html, "html.parser")
    #lista_diccionarios = []
    for anuncio in soup.find_all('div', class_='ad-preview'):   #Cada anuncio que contenga la clase 'ad-preview'.
        #time.sleep(10)
        script_json = anuncio.find('script', type='application/ld+json')   #Extraemos los datos.
        if script_json:
            data = json.loads(script_json.text)   #Cargamos los datos.

            piso = {   #Guardamos varios datos.
                "nombre": data.get("name"),
                "id": data.get("@id"),
                "detalle_url": "https://www.pisos.com" + data.get("url"),
                "latitud": data["geo"]["latitude"],
                "longitud": data["geo"]["longitude"],
                "localidad": data["address"]["addressLocality"],
                "region": data["address"]["addressRegion"]
            }
            precio_tag = anuncio.find('span', class_='ad-preview__price')   #Busca el precio, que esta fuera del script.
            piso["precio"] = precio_tag.text.strip() if precio_tag else None   #Se limpia con strip(), y si no se encuentra nada se pone None.

            piso = recoger_datos_piso(piso["detalle_url"], piso)

            lista_diccionarios.append(piso)
    return lista_diccionarios


def recoger_datos_piso(url_piso, diccionario):
    html1 = urlopen(url_piso, timeout=30)

    # Parseamos el html.
    soup1 = BeautifulSoup(html1, "html.parser")

    features_container = soup1.find("div", class_="features-container")

    if features_container:
        # Buscar todos los bloques de característica
        for feature in features_container.find_all("div", class_="features__feature"):
            label_tag = feature.find("span", class_="features__label")
            value_tag = feature.find("span", class_="features__value")

            if label_tag:
                label = label_tag.text.strip()
                value = value_tag.text.strip() if value_tag else True  # True si no hay value, solo etiqueta
                diccionario[label] = value
    return diccionario

# Funciona a la perfeccion. :)
#print(recoger_datos_web("https://www.pisos.com/venta/pisos-alicante/", []))
```
